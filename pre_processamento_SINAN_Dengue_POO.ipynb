**Resumo da classificação:**

As características do conjunto de dados são predominantemente nominais, incluindo variáveis categóricas e códigos de identificação. Também estão presentes características ordinais, discretas e temporais, como as relacionadas à escolaridade, idade, semanas epidemiológicas e datas dos registros. Não foram identificadas características quantitativas contínuas entre as variáveis analisadas.

```python
# Classe para realizar o pré-processamento dos dados do SINAN

class AnaliseSINAN:

    def __init__(self, dados):
        # Guarda uma cópia dos dados originais
        self.dados_originais = dados.copy()

        # Cria uma cópia para realizar o pré-processamento
        self.dados = dados.copy()

    def verificar_dados(self):
        print("DIMENSÕES DA BASE")
        print("Quantidade de registros:", self.dados.shape[0])
        print("Quantidade de colunas:", self.dados.shape[1])

        print("\nTIPOS DAS VARIÁVEIS")
        print(self.dados.dtypes)

    def verificar_ausentes(self):
        print("VALORES AUSENTES POR COLUNA")

        ausentes = self.dados.isnull().sum()

        print(ausentes[ausentes > 0])

        print("\nTotal de valores ausentes:")
        print(self.dados.isnull().sum().sum())

    def verificar_duplicados(self):
        duplicados = self.dados.duplicated().sum()

        print("Quantidade de registros duplicados:", duplicados)

    def identificar_constantes(self):
        constantes = []

        for coluna in self.dados.columns:
            if self.dados[coluna].nunique(dropna=False) == 1:
                constantes.append(coluna)

        print("Colunas com apenas um valor distinto:")

        if constantes:
            print(constantes)
        else:
            print("Nenhuma coluna constante identificada.")

        return constantes

    def pre_processar(self, remover_constantes=False):
        # Remove colunas completamente duplicadas
        colunas_duplicadas = self.dados.columns[
            self.dados.columns.duplicated()
        ]

        self.dados = self.dados.loc[
            :, ~self.dados.columns.duplicated()
        ].copy()

        # Remove registros exatamente duplicados
        self.dados = self.dados.drop_duplicates().copy()

        # Identifica colunas constantes
        constantes = self.identificar_constantes()

        # Remove colunas constantes somente se solicitado
        if remover_constantes and constantes:
            self.dados = self.dados.drop(
                columns=constantes
            ).copy()

        print("PRÉ-PROCESSAMENTO CONCLUÍDO")

        print("Colunas duplicadas removidas:", len(colunas_duplicadas))

        print(
            "Quantidade de registros após o processamento:",
            self.dados.shape[0]
        )

        print(
            "Quantidade de colunas após o processamento:",
            self.dados.shape[1]
        )

    def resumo_final(self):
        print("RESUMO FINAL DOS DADOS")

        print("Dimensões:", self.dados.shape)

        print(
            "\nTotal de valores ausentes:",
            self.dados.isnull().sum().sum()
        )

        print(
            "Total de registros duplicados:",
            self.dados.duplicated().sum()
        )
```

## Pré-processamento de dados utilizando POO

Nesta atividade, foi desenvolvida a classe `AnaliseSINAN`, com o objetivo de organizar e reutilizar as etapas de verificação e pré-processamento da base de dados do SINAN Dengue 2026.

A classe utiliza os conceitos de POO estudados em aula, incluindo classe, objeto, atributos, métodos, `self` e o método construtor `__init__()`.

Foram implementados métodos para verificar as dimensões e os tipos das variáveis, identificar valores ausentes, verificar registros duplicados e identificar colunas com apenas um valor distinto.

O método `pre_processar()` realiza a remoção de colunas com nomes duplicados e de registros exatamente duplicados. A remoção de colunas constantes é opcional, pois essas variáveis devem ser avaliadas conforme o objetivo da análise.

A classe também preserva uma cópia dos dados originais, permitindo manter os registros anteriores às transformações.

Dessa forma, o código fica organizado em métodos reutilizáveis, facilitando sua aplicação a outras bases de dados e períodos, desde que apresentem uma estrutura compatível.

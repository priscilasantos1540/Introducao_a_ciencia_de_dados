```python
# Importando a biblioteca Pandas
import pandas as pd
url = "https://media.githubusercontent.com/media/priscilasantos1540/Introducao_a_ciencia_de_dados/refs/heads/main/dados/DENGBR26.csv?token=B3AFPGJNJCWYAZWGCKGO663K23UD6"
df = pd.read_csv(url)

# Conferindo as dimensões da base
print("DIMENSÕES DA BASE")
print("Quantidade de registros:", df.shape[0])
print("Quantidade de colunas:", df.shape[1])

# Visualizando os cinco primeiros registros
print("\nPRIMEIROS CINCO REGISTROS")
display(df.head())
```

```python
# Importando a biblioteca para criar gráficos
import matplotlib.pyplot as plt

# Contando as notificações por semana epidemiológica
notificacoes_semana = df["SEM_NOT"].value_counts().sort_index()

# Criando o gráfico de barras
plt.figure(figsize=(12, 5))

plt.bar(
    notificacoes_semana.index.astype(str),
    notificacoes_semana.values
)

# Personalizando o gráfico
plt.title("Quantidade de notificações de dengue por semana epidemiológica - 2026")
plt.xlabel("Semana epidemiológica")
plt.ylabel("Quantidade de notificações")
plt.xticks(rotation=90)

# Ajustando a apresentação
plt.tight_layout()

# Exibindo o gráfico
plt.show()
```

### 5. Mapa de calor — Correlação entre sintomas de dengue

**Objetivo:** visualizar a correlação entre as variáveis relacionadas aos sintomas de dengue presentes na base de dados do SINAN 2026.

**Técnica utilizada:** mapa de calor (*Heatmap*), construído com as bibliotecas Seaborn e Matplotlib.

**Interpretação:** o gráfico apresenta uma matriz de correlação entre as variáveis FEBRE, MIALGIA, CEFALEIA, VOMITO e NAUSEA. Os valores numéricos e as cores permitem observar a intensidade e a direção da associação entre os códigos dessas variáveis. Valores próximos de 1 indicam associação positiva, valores próximos de -1 indicam associação negativa e valores próximos de 0 indicam pouca associação linear.

**Observação:** as variáveis de sintomas são registradas por meio de códigos, e não como medidas contínuas. Por isso, a interpretação deve considerar a codificação utilizada no SINAN. Correlação não significa causalidade, e os resultados representam os registros disponíveis na base analisada.

```python
# Importando as bibliotecas necessárias para o Dashboard
import matplotlib.pyplot as plt
import seaborn as sns

# Preparando os dados para os gráficos
anos_nascimento = df["ANO_NASC"].dropna()

notificacoes_semana = df["SEM_NOT"].value_counts().sort_index()

dados_scatter = df[["ANO_NASC", "SEM_NOT"]].dropna().copy()
dados_scatter["SEMANA"] = dados_scatter["SEM_NOT"] % 100

variaveis_sintomas = [
    "FEBRE",
    "MIALGIA",
    "CEFALEIA",
    "VOMITO",
    "NAUSEA"
]

correlacao = df[variaveis_sintomas].corr()

# Criando o painel com quatro gráficos
fig, eixos = plt.subplots(2, 2, figsize=(16, 11))

# Gráfico 1: Histograma
eixos[0, 0].hist(
    anos_nascimento,
    bins=30,
    edgecolor="black"
)
eixos[0, 0].set_title("Distribuição por ano de nascimento")
eixos[0, 0].set_xlabel("Ano de nascimento")
eixos[0, 0].set_ylabel("Quantidade de registros")

# Gráfico 2: Notificações por semana
eixos[0, 1].bar(
    notificacoes_semana.index.astype(str),
    notificacoes_semana.values
)
eixos[0, 1].set_title("Notificações por semana epidemiológica")
eixos[0, 1].set_xlabel("Semana epidemiológica")
eixos[0, 1].set_ylabel("Quantidade de registros")
eixos[0, 1].tick_params(axis="x", rotation=90)

# Gráfico 3: Dispersão
sns.scatterplot(
    data=dados_scatter,
    x="SEMANA",
    y="ANO_NASC",
    alpha=0.3,
    ax=eixos[1, 0]
)
eixos[1, 0].set_title("Ano de nascimento e semana de notificação")
eixos[1, 0].set_xlabel("Semana epidemiológica")
eixos[1, 0].set_ylabel("Ano de nascimento")

# Gráfico 4: Mapa de calor
sns.heatmap(
    correlacao,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f",
    ax=eixos[1, 1]
)
eixos[1, 1].set_title("Correlação entre sintomas")

# Ajustando o espaço entre os gráficos
fig.suptitle(
    "Dashboard — Visualização avançada dos dados de dengue em 2026",
    fontsize=15
)
plt.tight_layout()

# Exibindo o painel
plt.show()
```

### 6. Dashboard — Visão geral dos registros de dengue

**Objetivo:** reunir diferentes técnicas de visualização em um único painel para facilitar a análise exploratória dos dados.

**Técnicas utilizadas:** histograma, gráfico de barras, gráfico de dispersão e mapa de calor, utilizando Matplotlib e Seaborn.

**Interpretação:** o painel permite observar diferentes aspectos da base, como a distribuição dos anos de nascimento, a quantidade de notificações por semana epidemiológica, a distribuição dos registros entre duas variáveis e a correlação entre os códigos dos sintomas.

**Observação:** cada gráfico apresenta uma perspectiva específica dos registros. As interpretações devem considerar os valores ausentes, a codificação das variáveis e as limitações da base.

```python
# Importando a biblioteca Plotly Express
import plotly.express as px

# Preparando os dados para o gráfico interativo
dados_plotly = df[["ANO_NASC", "SEM_NOT"]].dropna().copy()

# Extraindo o número da semana epidemiológica
dados_plotly["SEMANA"] = dados_plotly["SEM_NOT"] % 100

# Criando o gráfico interativo
fig = px.scatter(
    dados_plotly,
    x="SEMANA",
    y="ANO_NASC",
    title="Visualização interativa: ano de nascimento e semana de notificação",
    labels={
        "SEMANA": "Semana epidemiológica",
        "ANO_NASC": "Ano de nascimento"
    },
    opacity=0.4,
    hover_data=["SEM_NOT"]
)

# Exibindo o gráfico
fig.show()
```
